# Active-learning Gaussian process regression

This notebook implements the active-learning workflow used for carrier-mobility
prediction: a 50-sample initial labeled set, iterative LightGBM feature screening,
Expected Improvement (EI) acquisition of 20 samples per round, and GPR evaluation
on a fixed 50-sample test set.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().name == "notebooks" else Path.cwd().resolve()
sys.path.insert(0, str(PROJECT_ROOT))

from src.active_learning import ActiveLearningConfig, run_active_learning
```

```python
config = ActiveLearningConfig(
    seed=2,
    init_size=50,
    test_size=50,
    select_per_iter=20,
    stop_threshold=0.01,
    patience=2,
    top_n_features=10,
    corr_threshold=0.95,
    gpr_restarts=20,
)
```

```python
results = run_active_learning(
    feature_csv=PROJECT_ROOT / "data" / "FVectors_MMS.csv",
    target_csv=PROJECT_ROOT / "data" / "N_type_carrier.csv",
    candidate_csv=PROJECT_ROOT / "data" / "candidate.csv",
    output_dir=PROJECT_ROOT / "results",
    config=config,
)

results["summary"]
```
